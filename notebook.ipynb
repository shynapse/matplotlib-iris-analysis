## Dataset

```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv(".\iris.csv")
```

## Petal length histogram

```python
plt.figure(figsize=(10,6))

for flower in df["variety"].unique():
    data = df[df["variety"] == flower]
    plt.hist(data["petal.length"],alpha=0.5,label=flower)

plt.title("Petal Length Histogram")
plt.xlabel("Petal Length")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

## Mean feature for each flower

```python
means = df.groupby("variety").mean()

means.T.plot(figsize=(10,6))

plt.title("Mean Features For Each Flower")
plt.ylabel("Mean Value")
plt.show()
```

## Avrage petal length by species

```python
mean_petal = df.groupby("variety")["petal.length"].mean()

plt.figure(figsize=(8,5))

plt.bar(mean_petal.index,mean_petal.values)

plt.title("Mean Petal Length")
plt.xlabel("Species")
plt.ylabel("Length")
plt.show()
```

## Species Distrbution

```python
flower_count = df["variety"].value_counts()

plt.pie(flower_count.values,labels=flower_count.index,autopct="%1.1f%%")
plt.title("Flower Distribution")
plt.show()
```

## Petal length vs width in scatter

```python
plt.figure(figsize=(8,6))

for flower in df["variety"].unique():

    subset = df[df["variety"] == flower]

    plt.scatter(subset["petal.length"],subset["petal.width"],label=flower)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Scatter Plot")

plt.legend()
plt.show()
```

## Subplot histogram for every species

```python
fig, axes = plt.subplots(2,2,figsize=(12,8))
features = [
    "sepal.length",
    "sepal.width",
    "petal.length",
    "petal.width"]

for ax,feature in zip(axes.flatten(),features):

    ax.hist(df[feature])
    ax.set_title(feature)

plt.tight_layout()
plt.show()
```

## Boxplot of iris features

```python
plt.boxplot([
    df["sepal.length"],
    df["sepal.width"],
    df["petal.length"],
    df["petal.width"]])

plt.xticks([1,2,3,4],["sepal.length","sepal.width","petal.length","petal.width"])
plt.ylabel("cm")
plt.grid(True)
plt.show()
```
